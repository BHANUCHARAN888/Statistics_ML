### <u>Margin of Error</u>
- Confidence Interval = Estimate ± Margin of Error
### <u>Hypothesis Testing</u>
- using sample data to decide whether there is enough evidence for a claim about a population.
- **Null Hypothesis — H₀**: The default/baseline assumption.
- **Alternative Hypothesis — H₁ / Hₐ**: What we are looking for evidence against H₀.
### <u>Significance Level (α)</u>
- The significance level is the <u>threshold</u> we decide before the test for how much evidence we need before rejecting the null hypothesis.
  - suppose, α = 0.05, p-value = 0.02
  - α is threshold level, p-value is evidence of statement
  - `α < p-value ---> fails to reject H₀` or `α ≥ p-value ---> reject H₀`
  - here α(0.05) ≥ p-value(0.02) --->reject H₀
### <u>p-value</u>
- p-value tells us how unusual our observed result would be if H₀ were actually true.
### <u>Type I & Type II Errors</u>
- **Type I Error — False Positive**: We reject H₀ even though H₀ is actually true.
- **Type II Error — False Negative**: We fail to reject H₀ even though H₀ is actually false.
### <u>Statistical Significance</u>
- p ≤ α
  - → statistically significant
  - → reject H₀
- p > α
  - → not statistically significant
  - → fail to reject H₀
### <u>Covariance</u>
- Covariance tells us the direction in which two variables move together.
  - Both increase together     → Positive covariance
  - One increases, other falls  → Negative covariance
  - No clear joint movement    → Around zero
  - Covariance depends on the units/scales of the variables, so its number is difficult to interpret directly.
### <u>Correlation</u>
- Correlation tells us both the direction AND strength of a linear relationship.
- It is standardized, so it always lies between:
- -1 ≤ r ≤ +1
  - +1  → Perfect positive relationship
  - 0  → No linear relationship
  - -1  → Perfect negative relationship
