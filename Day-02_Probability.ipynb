### <u>Probability Fundamentals</u>
- Probability tells us how likely an event is to happen.
- `0 (Impossible)  ─────────────────────────────── 1(Certain)`
  - Examples:
     * Probability of the sun rising tomorrow → very close to 1
     * Probability of rolling a 7 with a normal die → 0
     * Probability of getting heads from a fair coin → 0.5
- P(A)= $\frac{Number of favourable outcomes}{Total number of outcomes}$

### <u>Complement</u>
- **Complement concept**: P(~A) = 1 - P(A)
- **Conditional Probability**: P(A|B) = $\frac{P(A \cap B)}{P(B)}$
  - Example:There are 20 students:
     * 12 are boys
     * 8 are girls
     * 6 boys wear glasses
     * The selected student is a boy.Now, what is the probability that the student wears glasses?
         * P(glass | boys) =  $\frac{P(glass \cap Boys)}{P(Boys)}$
         * P(glass | boys) = $\frac{6}{12}$ ~ 0.5 ~ 50%

### <u>Independence</u>
- **Independence**:
     - Two events are independent if:
          * Knowing that one event happened does NOT change the probability of the other event.
     -  **Conditional probability**
          * If <u>A and B are independent</u>: P(A | B) = P(A)
     - Example:A bag contains 5 red and 5 blue balls,
          * You draw one ball without replacement.The first ball is red.
          * Now, the probability of drawing a red ball on the second draw is?
          * 1st red drawn --> 5/10 ~ 50%
          * 2nd red drawn --> 4/9 ~ 44.44%
          * finally both are not independent, they are dependent.
     - **Formal Independence**: P(A∩B)=P(A)×P(B)

### <u>Bayes Theorem</u>
- **Bayes Theorem**
     * Bayes Theorem is used to find the probability of something given that we already observed some evidence.
     * There are 100 emails:
         * 20 are actually spam.
         * A detector flags 25 emails as spam.
         * Out of those 25 flagged emails, 15 are actually spam.
         * If an email is flagged as spam, what is the probability that it is actually spam?
         * P(Spam | Flagged) = (15/25) * 100 = 60%
         * the probability that the email is actually spam given that it was flagged is 60%.
- **Bayes Formula**: P(A|B)=$\frac{P(B|A)\times P(A)}{P(B)}$
    * where,
         * P(A|B) → Probability of A given B
         * P(B|A) → Probability of B given A
         * P(A)   → Prior probability of A
         * P(B)   → Probability of B
      
</br> 

- One more Example:
    * A company has 1,000 customers:
    * 100 customers are fraudulent.
    * 900 customers are legitimate.
    * The fraud detection system correctly flags 80 of the 100 fraudulent customers.
    * It also incorrectly flags 90 legitimate customers as fraudulent.
    * If a customer is flagged as fraudulent, what is the probability that the customer is actually fraudulent?
    * P(Fraud | Flag) = $\frac{(P(Flag | Fraud)) * P(Fraud)}{P(Flag)}$
    * Flagged customers = 80 fraudulent + 90 legitimate = 170
    * Actually fraudulent among flagged = 80
    * P(Fraud | Flagged) = 80/170 = **47.06%**

### <u>Expected Value</u>
- Expected Value (EV) tells us the average outcome we would expect if we repeated a random experiment many times.
- **Formula** = EV=∑(Outcome×Probability)
- Example: Suppose,
     - 20% chance → win ₹100
     - 50% chance → win ₹20
     - 30% chance → lose ₹50
     - EV=(0.2×100)+(0.5×20)+(0.3×(−50))
     - EV=20+10−15 = **₹15**
