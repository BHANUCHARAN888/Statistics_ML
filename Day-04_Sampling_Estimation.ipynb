### <u>Population v/s Sample</u>
- If a company has 100,000 customers and we randomly select 1,000 customers to study their purchasing behavior:
   - 100,000 ---> Population
   - 1000 ---> Sample, to get clarity/estimation of rest.

### <u>Sampling Methods</u>
-This is simply how we choose the sample from the population
   - We mainly need 4 methods for our ML/statistics level:
     - **Simple Random Sampling**: Every individual in the population has an equal chance of being selected.
       - Ex: Population = 10,000 customers, Customer IDs: 1-10000, Randomly select 500---> Sample
     - **Systematic Sampling**: Instead of randomly choosing every person, we choose every fixed interval from the population.
       - Ex: Suppose we have 1,000 customers and want a sample of 100.
       - k = $\frac{1000}{100}$ = 10, every 10th customer start from anywhere, but 10th one we consider.  
     - **Stratified Sampling**: We divide the population into meaningful groups (strata), then randomly select people from each group.
       - Ex: A college has 5,000 students,
         - 1st Year → 1,500
         - 2nd Year → 1,300
         - 3rd Year → 1,200
         - 4th Year → 1,000
         - Randomly select students from each year.
     - **Cluster Sampling**: We divide the population into groups (clusters), then randomly select entire groups instead of selecting individuals from every group.

### <u>Sampling Distribution</u>
- A sampling distribution is the distribution of a statistic (like the mean) calculated from many samples.
- Example: (Suppose the population has 10,000 students) ---> Population,
  - Sample 1 → calculate mean → 70
  - Sample 2 → calculate mean → 72
  - Sample 3 → calculate mean → 69...
  - ... collect those mean and make distribution to Estimate the population.

### <u>Point Estimation</u>
- Using one value from a sample to estimate the corresponding value of the population.
### <u>Confidence Intervals</u>
- Sample mean → one value used to estimate the population mean.
- `But one value doesn't tell us how uncertain our estimate is.`
- Confidence Interval = a range of likely values
- mean = 500, margin of error: ± 30 ----------> [470, 530]
