### <u>Distributions</u>
- A distribution describes how values are spread and how frequently they occur.
- Example: 1, 2, 1, 3, 2, 1, 4, 2, 1, 3
- (Value)--->(Frequency):  1->4, 2->3, 3->2, 4->1
### <u>Discrete vs Continuous</u>
- Discrete → values are countable.
- Example: From the dataset,
   - Number of customers = 100
   - Number of emails = 25
   - Number of defects = 3
   - Number of children = 2
     
</br>

- Continuous → values can take any value within a range.
- Example:
   - Height = 172.53 cm
   - Weight = 68.72 kg
   - Temperature = 31.47°C
   - Time = 2.384 seconds 

### <u>Bernoulli Distribution</u>
- A Bernoulli distribution models an experiment/trail that has exactly two possible outcomes.
- Example: Yes/No, True/False
- Formula:$P(X=x) = p^x (1-p)^{1-x}$

     - p = probability of success
     - 1-p = probability of failure
     - where, X = 1 → Success, X = 0 → Failure
- **Bernouli v/s Binomial**:
     - Customer buys(yes / no)? → Bernoulli
     - How many buy(in a group of customers)? → Binomial

### <u>Binomial Distribution</u>
- Binomial → MULTIPLE Bernoulli trials
- Four condition that Binomial distribution follows,
     - Fixed number of trials
     - Each trial has only 2 outcomes
     - Trials are independent
     - Probability of success stays the same
- Formula: $P(X=k) = \binom{n}{k} p^k (1-p)^{n-k}$
- where,
     - n = total number of trials
     - k = number of successes we want
     - p = probability of success
     - 1-p = probability of failure
- Example: Flip a fair coin 3 times. What is the probability of getting exactly 2 Heads?
     - $P(X=2) = \binom{3}{2} p^2 (1-p)^{3-2}$
     - where p = 0.5, 1-p = 0.5
     -  $\binom{3}{2}$ = $\frac{3!}{2!1!}$ = 3
     -  $P(X=2) = 3(0.5)^2(0.5)$ = **0.375**
     -  0.375 * 100 = **37.5%**

### <u>Poisson Distribution</u>
- Poisson Distribution tells us the probability of a certain number of events happening within a fixed interval of time or space.
- $P(X = k) = \frac{e^{-\lambda} \lambda^k}{k!}$
     - λ (lambda) = average number of events in the interval
     - k          = number of events we want
     - e          ≈ 2.718
     - k!         = factorial of k
  
</br>

- Example: Suppose a customer-support system receives an average of 3 calls per hour.
     - What is the probability of receiving exactly 2 calls in one hour?
     - $P(X = 2) = \frac{2.718^{-3} 3^2}{2!}$
     - = $\frac{(0.0498)(9)}{2}$ = 0.2241 = **22.41%**
     - `So there's approximately a 22.4% probability of exactly 2 calls in one hour.`

### <u>Uniform Distribution</u>
- Uniform Distribution means every possible value (or interval) has the same probability.
- **Discrete uniform distribution**:
    - A fair die has:
        - 1 → 1/6
        - 2 → 1/6
        - 3 → 1/6
        - 4 → 1/6
        - 5 → 1/6
        - 6 → 1/6
        - `Every outcome has the same probability.`
- **Continuous Uniform Distribution**:
- P(a ≤ X ≤ b) = $\frac{(b-a)}{(upper-lower)}$
- Example: A random number is uniformly distributed between 0 and 20.What is the probability that the number falls between 5 and 15?
    - P(5 ≤ X ≤ 15) =  $\frac{(15-5)}{(20-0)}$
    - = $\frac{10}{20}$ = `50% probability that no. falls in btw 5 and 15 within the interval 0 and 20.`

### <u>Normal Distribution</u>
- A Normal Distribution is a distribution where values are concentrated around the middle, and fewer values occur as we move away from the middle.
- It looks like a bell shape:
- Example:
     - <img src="https://homework.study.com/cimages/multimages/16/normal2847021515433714845.png" width="300" height="200">
- **68–95–99.7 Rule**:
     - 1σ → ~68% of data
     - 2σ → ~95% of data
     - 3σ → ~99.7% of data
- **Mean = Median = Mode**: In a perfectly normal distribution, all three meet at the center.
- **Z-score**: It tells us, how many standard deviations a value is away from the mean.
     - Formula: $\frac{(x - μ)}{σ}$
     - Ex 1: mean = 70, std = 10, data point = 90
     - => $\frac{(90 - 70)}{10}$ => $\frac{20}{10}$ => **+2** deviation from the mean to data point.(Above the mean(+2σ))
</br>

     - Ex 2: mean = 50, std = 10, data point = 30
     - => $\frac{(30 - 50)}{10}$ => $\frac{-20}{10}$ => **-2** deviation from the mean to data point.(below the mean(-2σ))

### <u>Central Limit Theorem</u>
- Many samples → calculate their means → sample means tend toward a Normal Distribution.

### <u>Skewness</u>
- Skewness tells us whether a distribution is symmetric or has a longer tail on one side.
- **Symmetric**: Values are balanced around the center. (Skewness ~ 0)
- **Right Skew / Positive Skew**: The distribution has a long tail toward the right.(Mean > Median)
- **Left Skew / Negative Skew**: The distribution has a long tail toward the left.(Mean < Median)
